# Step 1: Data understanding

Load the source data and review its structure, fields, sample records, and basic distributions. Keep the source file unchanged.

```python
from pathlib import Path
import pandas as pd

raw_path = Path("../data/raw/ecommerce_delivery_analytics.csv")
df = pd.read_csv(raw_path)
print(f"Rows: {len(df):,} | Columns: {len(df.columns)}")
display(df.head())
```

```python
display(df.dtypes.rename("dtype").to_frame())
display(df.isna().sum().rename("missing").to_frame())
print(f"Duplicate rows: {df.duplicated().sum():,}")
display(df.describe(include="all").T)
```
